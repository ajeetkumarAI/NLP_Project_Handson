```
# Importing essential libraries
import numpy as np
import pandas as pd
```

```
# Loading the dataset
df = pd.read_csv('/content/Restaurant_Reviews.tsv', delimiter='\t', quoting=3)
```

# **Data Preprocessing**

```
# Importing essential libraries for performing Natural Language Processing on 'Restaurant_Reviews.tsv' dataset
import nltk
import re
nltk.download('stopwords')
from nltk.corpus import stopwords
from nltk.stem.porter import PorterStemmer
```

```
# Cleaning the reviews
corpus = []
for i in range(0,1000):

  # Cleaning special character from the reviews
  review = re.sub(pattern='[^a-zA-Z]',repl=' ', string=df['Review'][i])

  # Converting the entire review into lower case
  review = review.lower()

  # Tokenizing the review by words
  review_words = review.split()

  # Removing the stop words
  review_words = [word for word in review_words if not word in set(stopwords.words('english'))]

  # Stemming the words
  ps = PorterStemmer()
  review = [ps.stem(word) for word in review_words]

  # Joining the stemmed words
  review = ' '.join(review)

  # Creating a corpus
  corpus.append(review)
```

```
corpus[0:10]
```

```
# Creating the Bag of Words model
from sklearn.feature_extraction.text import CountVectorizer
cv = CountVectorizer(max_features=1500)
X = cv.fit_transform(corpus).toarray()
y = df.iloc[:, 1].values
```

# **Model Building**

```
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size = 0.20, random_state = 0)
```

```
# Fitting Naive Bayes to the Training set
from sklearn.naive_bayes import MultinomialNB
classifier = MultinomialNB()
classifier.fit(X_train, y_train)
```

```
# Predicting the Test set results
y_pred = classifier.predict(X_test)
```

```
# Accuracy, Precision and Recall
from sklearn.metrics import accuracy_score
from sklearn.metrics import precision_score
from sklearn.metrics import recall_score
score1 = accuracy_score(y_test,y_pred)
score2 = precision_score(y_test,y_pred)
score3= recall_score(y_test,y_pred)
print("---- Scores ----")
print("Accuracy score is: {}%".format(round(score1*100,2)))
print("Precision score is: {}".format(round(score2,2)))
print("Recall score is: {}".format(round(score3,2)))
```

```
# Making the Confusion Matrix
from sklearn.metrics import confusion_matrix
cm = confusion_matrix(y_test, y_pred)
```

```
cm
```

```
# Plotting the confusion matrix
import matplotlib.pyplot as plt
import seaborn as sns
%matplotlib inline

plt.figure(figsize = (10,6))
sns.heatmap(cm, annot=True, cmap="YlGnBu", xticklabels=['Negative', 'Positive'], yticklabels=['Negative', 'Positive'])
plt.xlabel('Predicted values')
plt.ylabel('Actual values')
```

```
# Hyperparameter tuning the Naive Bayes Classifier
best_accuracy = 0.0
alpha_val = 0.0
for i in np.arange(0.1,1.1,0.1):
  temp_classifier = MultinomialNB(alpha=i)
  temp_classifier.fit(X_train, y_train)
  temp_y_pred = temp_classifier.predict(X_test)
  score = accuracy_score(y_test, temp_y_pred)
  print("Accuracy score for alpha={} is: {}%".format(round(i,1), round(score*100,2)))
  if score>best_accuracy:
    best_accuracy = score
    alpha_val = i
print('--------------------------------------------')
print('The best accuracy is {}% with alpha value as {}'.format(round(best_accuracy*100, 2), round(alpha_val,1)))
```

# **Predictions**

```
def predict_sentiment(sample_review):
  sample_review = re.sub(pattern='[^a-zA-Z]',repl=' ', string = sample_review)
  sample_review = sample_review.lower()
  sample_review_words = sample_review.split()
  sample_review_words = [word for word in sample_review_words if not word in set(stopwords.words('english'))]
  ps = PorterStemmer()
  final_review = [ps.stem(word) for word in sample_review_words]
  final_review = ' '.join(final_review)

  temp = cv.transform([final_review]).toarray()
  return classifier.predict(temp)
```

```
# Predicting values
sample_review = 'The food is really good here.'

if predict_sentiment(sample_review):
  print('This is a POSITIVE review.')
else:
  print('This is a NEGATIVE review!')
```

```
import numpy as np
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.model_selection import train_test_split
from sklearn.naive_bayes import MultinomialNB
from sklearn.metrics import accuracy_score, precision_score, recall_score, confusion_matrix
import matplotlib.pyplot as plt
import seaborn as sns
import re
from nltk.corpus import stopwords
from nltk.stem.porter import PorterStemmer

print("--- Implementing with TF-IDF ---")

# Creating the TF-IDF model
# Using max_features similar to CountVectorizer for consistency
tfidf_cv = TfidfVectorizer(max_features=1500)
X_tfidf = tfidf_cv.fit_transform(corpus).toarray()
y_tfidf = df.iloc[:, 1].values # Labels remain the same

# Splitting the dataset into the Training set and Test set
X_train_tfidf, X_test_tfidf, y_train_tfidf, y_test_tfidf = train_test_split(X_tfidf, y_tfidf, test_size = 0.20, random_state = 0)

# Fitting Naive Bayes to the Training set with TF-IDF
classifier_tfidf = MultinomialNB()
classifier_tfidf.fit(X_train_tfidf, y_train_tfidf)

# Predicting the Test set results
y_pred_tfidf = classifier_tfidf.predict(X_test_tfidf)

# Accuracy, Precision and Recall for TF-IDF model
score1_tfidf = accuracy_score(y_test_tfidf, y_pred_tfidf)
score2_tfidf = precision_score(y_test_tfidf, y_pred_tfidf)
score3_tfidf = recall_score(y_test_tfidf, y_pred_tfidf)
print("\n---- TF-IDF Model Scores ----")
print("Accuracy score is: {}%".format(round(score1_tfidf*100,2)))
print("Precision score is: {}".format(round(score2_tfidf,2)))
print("Recall score is: {}".format(round(score3_tfidf,2)))

# Making the Confusion Matrix for TF-IDF model
cm_tfidf = confusion_matrix(y_test_tfidf, y_pred_tfidf)
print("\nConfusion Matrix (TF-IDF):")
print(cm_tfidf)

# Plotting the confusion matrix for TF-IDF
plt.figure(figsize = (10,6))
sns.heatmap(cm_tfidf, annot=True, cmap="YlGnBu", xticklabels=['Negative', 'Positive'], yticklabels=['Negative', 'Positive'])
plt.xlabel('Predicted values')
plt.ylabel('Actual values')
plt.title('Confusion Matrix for TF-IDF Model')
plt.show()

# Hyperparameter tuning for TF-IDF model
print("\n---- Hyperparameter tuning for TF-IDF Model ----")
best_accuracy_tfidf = 0.0
alpha_val_tfidf = 0.0
for i in np.arange(0.1,1.1,0.1):
  temp_classifier_tfidf = MultinomialNB(alpha=i)
  temp_classifier_tfidf.fit(X_train_tfidf, y_train_tfidf)
  temp_y_pred_tfidf = temp_classifier_tfidf.predict(X_test_tfidf)
  score_tfidf = accuracy_score(y_test_tfidf, temp_y_pred_tfidf)
  print("Accuracy score for alpha={} is: {}%".format(round(i,1), round(score_tfidf*100,2)))
  if score_tfidf > best_accuracy_tfidf:
    best_accuracy_tfidf = score_tfidf
    alpha_val_tfidf = i
print('--------------------------------------------')
print('The best accuracy for TF-IDF model is {}% with alpha value as {}'.format(round(best_accuracy_tfidf*100, 2), round(alpha_val_tfidf,1)))

# Update classifier with best alpha for TF-IDF
classifier_tfidf = MultinomialNB(alpha=alpha_val_tfidf)
classifier_tfidf.fit(X_train_tfidf, y_train_tfidf)


# Prediction function for TF-IDF model
print("\n---- Example Prediction with TF-IDF Model ----")
sample_review_for_tfidf = 'The food is really good here.'

# Clean and transform the sample review using TF-IDF vectorizer
processed_sample_review_for_tfidf = re.sub(pattern='[^a-zA-Z]',repl=' ', string = sample_review_for_tfidf)
processed_sample_review_for_tfidf = processed_sample_review_for_tfidf.lower()
processed_sample_review_words_for_tfidf = processed_sample_review_for_tfidf.split()
processed_sample_review_words_for_tfidf = [word for word in processed_sample_review_words_for_tfidf if not word in set(stopwords.words('english'))]
ps = PorterStemmer()
final_review_for_tfidf = [ps.stem(word) for word in processed_sample_review_words_for_tfidf]
final_review_for_tfidf = ' '.join(final_review_for_tfidf)

temp_tfidf = tfidf_cv.transform([final_review_for_tfidf]).toarray()
prediction_tfidf = classifier_tfidf.predict(temp_tfidf)

if prediction_tfidf[0] == 1:
  print("'{}' is a POSITIVE review (TF-IDF model).".format(sample_review_for_tfidf))
else:
  print("'{}' is a NEGATIVE review (TF-IDF model).".format(sample_review_for_tfidf))
```
