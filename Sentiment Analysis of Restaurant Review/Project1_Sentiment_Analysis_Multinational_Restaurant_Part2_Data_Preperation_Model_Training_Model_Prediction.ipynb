````python
# Initialize clean corpus collection list
corpus = []

# Loop over all 1000 dataset reviews and process systematically
for i in range(0, 1000):
    # Step 1: Clean non-alphabetic characters
    review_without_special_char = re.sub(pattern="[^a-zA-Z]", repl=" ", string=df['Review'][i])

    # Step 2: Standardize letter casing
    review_normalization = review_without_special_char.lower()

    # Step 3: Remove custom stopwords
    review_without_stopwords = [word for word in review_normalization.split() if word not in custom_stopwords]

    # Step 4: Reduce terms to root structure
    ps = PorterStemmer()
    review_with_stem_word = [ps.stem(word) for word in review_without_stopwords]

    # Step 5: Compile list into processed text structure
    review = " ".join(review_with_stem_word)

    # Step 6: Save processed text to corpus
    corpus.append(review)
````

````python
# --- Insights and Summary of Text Preprocessing ---

# This section focuses on preparing the raw text reviews for machine learning by applying a series of preprocessing techniques.
# The goal is to clean the text, standardize it, and reduce its dimensionality, which are crucial steps for effective text analysis.

# 1. Removal of Special Characters and Punctuation:
#    - Using regular expressions (`re.sub(pattern="[^a-zA-Z]", repl=" ", string=...)`), all characters that are not English alphabet letters
#      are replaced with spaces. This helps to remove noise such as numbers, symbols, and extra punctuation that do not contribute to sentiment analysis.

# 2. Text Normalization (Lowercasing):
#    - All words are converted to lowercase (`.lower()`). This ensures that words like 'Wow', 'wow', and 'WOW' are treated as the same word,
#      preventing the model from learning redundant features and improving consistency.

# 3. Stopword Removal:
#    - Common words (stopwords like 'the', 'a', 'is') that carry little to no semantic meaning for sentiment analysis are removed.
#    - A custom stopword list (`custom_stopwords`) was created by modifying the NLTK English stopwords to *retain* negative words
#      (e.g., 'not', 'no', 'don't'). This is a crucial step to preserve the sentiment polarity of reviews, as these words directly impact meaning.
#      For example, 'food was good' and 'food was not good' have opposite sentiments, and removing 'not' would lose this distinction.

# 4. Stemming:
#    - Using NLTK's `PorterStemmer`, words are reduced to their root form (e.g., 'loved' becomes 'love', 'eating' becomes 'eat').
#    - This technique (or lemmatization, a more linguistically informed approach) helps in reducing the vocabulary size and treating
#      different inflections of a word as a single term, further reducing dimensionality and improving model generalization.

# 5. Corpus Creation:
#    - After applying all these steps, the cleaned and processed words for each review are joined back into a string.
#    - These processed review strings are then collected into a list called `corpus`. This `corpus` now represents the clean, preprocessed
#      text data, ready for the next stage: feature extraction (e.g., Bag-of-Words or TF-IDF) to convert text into numerical format for machine learning models.
#    - The `corpus` effectively transforms 1000 raw reviews into a structured format suitable for building a sentiment classifier.
````

````python
from sklearn.feature_extraction.text import CountVectorizer

# --- Text Feature Extraction using CountVectorizer ---
# CountVectorizer is a tool from scikit-learn that converts a collection of text documents
# into a matrix of token counts. It's a fundamental step in preparing text data for
# machine learning models, as models require numerical input.

# How it works:
# 1. Tokenization: Splits text into individual words (tokens).
# 2. Vocabulary Creation: Builds a dictionary of all unique words in the corpus.
# 3. Word Counting: Transforms text into a matrix where each row represents a document
#    and each column represents a word from the vocabulary, with cell values indicating
#    the frequency of that word in the document.

# The `max_features` parameter (used in the subsequent cell) is crucial for controlling
# the size of the vocabulary. If set, it instructs CountVectorizer to only consider the
# top 'n' words ordered by their frequency across the entire corpus. This helps in:
# - Reducing dimensionality: Limits the number of features (words), making the model
#   computationally less expensive and reducing the risk of overfitting.
# - Focusing on most relevant words: Often, the most frequent words are the most
#   informative for classification tasks.
````

This cell conceptually visualizes the word frequency count. In essence, `CountVectorizer` builds an internal vocabulary, assigning an index to each unique word. The `max_features` parameter then filters this vocabulary, retaining only the `N` most frequent words, which are often the most impactful for sentiment analysis. Words like 'wow', 'love', and 'place' appear frequently, indicating their importance, while less frequent words like 'menu', 'late', and 'rick' would be discarded if `max_features` is set to a value that excludes them.

The comments in the cell above illustrate how `CountVectorizer` prioritizes words. When `max_features` is set (e.g., to 1500 as in `cv = CountVectorizer(max_features=1500)`), the `CountVectorizer` examines all unique words across the entire `corpus`.

It then ranks these words by their frequency of appearance. The `max_features` parameter ensures that only the `N` most frequent words are kept in the vocabulary. Words with lower frequencies (like 'menu', 'late', 'rick' in the example) might be discarded if `N` is exceeded. This process reduces the dimensionality of the feature space while retaining the most informative words for the model.

This is where the `CountVectorizer` truly transforms our text data into numerical features.

*   **`cv.fit_transform(corpus)`**: This method performs two main actions:
    1.  **`fit`**: It learns the vocabulary from the `corpus` (our list of cleaned reviews). Based on the `max_features=1500` setting, it identifies the 1500 most frequent words and creates a mapping (vocabulary) from these words to unique integer indices.
    2.  **`transform`**: It then converts each review in the `corpus` into a numerical vector. Each vector will have 1500 elements (corresponding to the 1500 selected words). The value at each position indicates the count of that specific word in the review.

*   **`.toarray()`**: The output of `fit_transform` is a sparse matrix (efficient for memory when most values are zero). `.toarray()` converts this sparse matrix into a dense NumPy array, which is a standard format for many machine learning algorithms. The resulting array `X` is our independent feature set, where each row is a review, and each column represents a word count.

This section evaluates the performance of our trained Gaussian Naive Bayes model using various metrics:

*   **`accuracy_score(y_test, y_pred)`**: Measures the proportion of correctly classified reviews out of the total number of reviews in the test set. An accuracy of 0.735 means the model correctly predicted the sentiment for 73.5% of the reviews.

*   **`confusion_matrix(y_test, y_pred)`**: Provides a detailed breakdown of correct and incorrect predictions. It's a table showing:
    *   True Positives (top-left): Reviews correctly predicted as positive.
    *   False Positives (top-right): Reviews incorrectly predicted as positive (Type I error).
    *   False Negatives (bottom-left): Reviews incorrectly predicted as negative (Type II error).
    *   True Negatives (bottom-right): Reviews correctly predicted as negative.

*   **`precision_score(y_test, y_pred)`**: Measures the accuracy of positive predictions. It answers: "Of all reviews predicted as positive, how many were actually positive?" A precision of 0.689 indicates that about 68.9% of the reviews the model classified as positive were indeed positive.

*   **`recall_score(y_test, y_pred)`**: Measures the ability of the model to find all the positive samples. It answers: "Of all the actual positive reviews, how many did the model correctly identify?" A recall of 0.883 suggests the model captured about 88.3% of all actual positive reviews.

*   **`f1_score(y_test, y_pred)`**: The F1-score is the harmonic mean of precision and recall. It provides a single score that balances both metrics, especially useful when there's an uneven class distribution. An F1-score of 0.774 represents a good balance between precision and recall for this model.
